# Explanation of the Dataset

The Adult Income dataset consists of demographic and employment-related information extracted from the 1994 
U.S. Census database. It features a combination of continuous and categorical attributes, totaling 14 features per 
instance, across 48,842 instances. The dataset's primary objective is to predict whether an individual earns more than 
$50,000 per year, a binary classification task with the classes being >50K and <=50K. Attributes include age, 
work class, education level, marital status, occupation, relationship status, race, sex, capital gain and loss, hours 
worked per week, and native country. These attributes provide a comprehensive view of each individual's socio-economic status,
employment characteristics, and demographic background.

# Explanation of the k-Nearest Neighbors (kNN) algorithm and its application

The k-Nearest Neighbors (kNN) algorithm classifies data points based on the closest neighbors in the feature space. 
Here is our approach for subsequent income classification task:

Principle: kNN predicts a data point's class by identifying the 'k' nearest neighbors in the feature space, 
using a majority vote for classification. It is a direct method that does not assume a specific data distribution.
We are utilizing Euclidean distance to measure the proximity between data points, an choice 
for calculating geometric distances in a multidimensional space.
Choice of k = 5: Selecting k=5 balances sensitivity to noise against the risk of overgeneralizing, aiming for a model 
that's neither too simple nor overly complex.
Key Considerations for kNN
Feature Scaling: Essential to ensure no attribute disproportionately influences the distance calculations due to its scale.
Dimensionality: Attention to feature relevance and scaling is crucial to mitigate the "curse of dimensionality" and maintain
model performance.

The k-Nearest Neighbors (kNN) algorithm is a straightforward method used for classifying data points by leveraging the 
data similarity between them in the feature space. This method is centered around the principle of predicting the class of a 
given data point based on the 'k' nearest neighbors to it within the feature space. This prediction is made through a 
majority vote among the neighbors, making kNN a direct and intuitive approach that does not require any assumptions 
about the data distribution.

For measuring the proximity between data points, we have selected Euclidean distance. 
This choice is based on its simplicity and effectiveness in calculating geometric distances in a multidimensional space, 
making it a good fit for our analysis. To balance the model sensitivity accounting for noise and generalization capability, 
we have chosen k=5. This selection helps to avoid overly simplistic or complex models, aiming for a balance that enhances
prediction accuracy while minimizing the risk of overfitting.

Proper feature scaling is crucial to avoid any single attribute from disproportionately affecting the distance 
calculations, ensuring that all features contribute equally to the model. Additionally, mitigating the 
"curse of dimensionality" is important for maintaining model performance. This involves applying scaling techniques
and feature selection to ensure the validty and accuracy of the kNN algorithm in our income classification task.

```python
# Visualising variable distributions
def diagnostic_plots(df, variable, target):
    # Define figure size.
    plt.figure(figsize=(20, 5))

    # Histogram
    plt.subplot(1, 3, 1)
    sns.histplot(df[variable], kde=True)
    plt.title('Histogram')
    
    # Boxplot
    plt.subplot(1, 3, 2)
    sns.boxplot(y=df[variable])
    plt.title('Boxplot')
    
    plt.show()
```

```python
# Data parsing
X = df1.drop(columns={"income"},axis=1)
y = df1["income"].values.reshape(-1,1) 
```

```python
# Data split: 80% Training and 20% Testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state = 11)
```

```python
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
# Scaling/ normalizing the features in the training and test sets to a range between 0 and 1 using MinMaxScaler
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.fit_transform(X_test)
```

```python
# Fititng the KNN classifier and setting the parameter 'k'equal to 5.
from sklearn.neighbors import KNeighborsClassifier

# Initialize the kNN classifier with k = 5
knn = KNeighborsClassifier(n_neighbors = 5, metric='minkowski', p=2)

# Train the k-nearest neighbors classifier on the training set
knn.fit(X_train, y_train)  # use .ravel() to convert y_train to a 1D array if needed
```

```python
# Classification predictions
y_pred = knn.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Calculate metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average = 'binary')
recall = recall_score(y_test, y_pred, average = 'binary')
f1 = f1_score(y_test, y_pred, average = 'binary')

# Print the metrics
print(f"Accuracy: {accuracy}")
print(f"Precision: {precision}")
print(f"Recall: {recall}")
print(f"F1 Score: {f1}")
```

```python
# Confusion Matrix
conf_matrix = confusion_matrix(y_test, y_pred)
print(f"Confusion Matrix:\n{conf_matrix}")
```

# Interpreting the Results with respect to the two research questions formed:

Accuracy in Predicting Income Categories: The accuracy score of 0.807 indicates that the kNN model correctly predicts the

Influence of Features on Income Levels: While the kNN algorithm used in this analysis does not explicitly quantify or rank

```python
from sklearn.model_selection import cross_val_score

# Perform 5-fold cross-validation
cv_scores = cross_val_score(knn, X, y, cv=5)  # Ensure y is a 1D array if needed using .ravel()

# Print the average accuracy across all folds
print(f"CV Average Score: {cv_scores.mean()}")
```
